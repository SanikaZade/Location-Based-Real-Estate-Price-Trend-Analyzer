```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load original file
df = pd.read_excel('housing RP.xlsx')  # Make sure the path is correct

# Select and rename columns
dataset = df[['title-style', 'subtitle-style', 'T_configurationStyle', 'T_blackText 2', 'sellerName-label']].copy()
dataset.columns = ['project_name', 'location', 'area_sqft', 'price_lakh', 'developer']

# Clean text columns
dataset['location'] = dataset['location'].astype(str).str.lower().str.strip()
dataset['developer'] = dataset['developer'].astype(str).str.lower().str.strip()

# Encode categorical values
le_loc = LabelEncoder()
le_dev = LabelEncoder()
dataset['location_encoded'] = le_loc.fit_transform(dataset['location'])
dataset['developer_encoded'] = le_dev.fit_transform(dataset['developer'])

# Convert area_sqft and price_lakh to numeric
dataset['area_sqft'] = pd.to_numeric(dataset['area_sqft'].astype(str).str.replace(',', ''), errors='coerce')
dataset['price_lakh'] = pd.to_numeric(dataset['price_lakh'].astype(str).str.replace(',', ''), errors='coerce')

# Drop rows where area or price is missing or zero
dataset = dataset.dropna(subset=['area_sqft', 'price_lakh'])
dataset = dataset[(dataset['area_sqft'] > 0) & (dataset['price_lakh'] > 0)]

# Create price per sqft column
dataset['price_per_sqft'] = (dataset['price_lakh'] * 100000) / dataset['area_sqft']

# Final cleaned dataset
final_data = dataset[['area_sqft', 'price_lakh', 'price_per_sqft', 'location_encoded', 'developer_encoded']]

# Save to Excel
final_data.to_excel('final_housing_data_cleaned.xlsx', index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load original file
df = pd.read_excel('housing RP.xlsx')

# Select and rename columns
dataset = df[['title-style', 'subtitle-style', 'T_configurationStyle', 'T_blackText 2', 'sellerName-label']].copy()
dataset.columns = ['project_name', 'location', 'area_sqft', 'price_lakh', 'developer']

# Extract numeric area
dataset['area_sqft'] = dataset['area_sqft'].astype(str).str.extract(r'(\d+\.?\d*)')
dataset['area_sqft'] = pd.to_numeric(dataset['area_sqft'], errors='coerce')

# Extract numeric price
dataset['price_lakh'] = dataset['price_lakh'].astype(str).str.replace(r'[^\d.]', '', regex=True)
dataset['price_lakh'] = dataset['price_lakh'].replace('', pd.NA)
dataset['price_lakh'] = pd.to_numeric(dataset['price_lakh'], errors='coerce')

# Drop bad rows
dataset = dataset.dropna(subset=['area_sqft', 'price_lakh'])
dataset = dataset[(dataset['area_sqft'] > 0) & (dataset['price_lakh'] > 0)]

# Compute price per sqft
dataset['price_per_sqft'] = (dataset['price_lakh'] * 100000) / dataset['area_sqft']

# Encode text
le_loc = LabelEncoder()
le_dev = LabelEncoder()
dataset['location'] = dataset['location'].astype(str).str.lower().str.strip()
dataset['developer'] = dataset['developer'].astype(str).str.lower().str.strip()
dataset['location_encoded'] = le_loc.fit_transform(dataset['location'])
dataset['developer_encoded'] = le_dev.fit_transform(dataset['developer'])

# Final cleaned dataset
final_data = dataset[['area_sqft', 'price_lakh', 'price_per_sqft', 'location_encoded', 'developer_encoded']]

# Confirm it has rows
print("Final row count:", final_data.shape[0])

# Save to Excel
final_data.to_excel('final_housing_data_cleaned.xlsx', index=False)
```
